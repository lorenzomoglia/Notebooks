Partiamo dall'inizio: cosa è una classe? E' un progetto, che dice alla macchina le caratteristiche che un oggetto dovrà avere, e le azioni che potrà compiere.  Le caratteristiche prendono il nome di attributi, e le azioni prendono il nome di metodi.   
  
Qua sotto definiamo una classe Animale, che prevede un attributo di classe "regno" popolato con animale.  
Il modo in cui noi creaiamo nuovi oggetti si chiama costruttore, e in python è una funzione che si chiama __init__(self, **),  e può essere solo uno per classe. Se vieni da un altro linguaggio questo può essere un fastidio, ma ci sono dei workaround di questa limitazione.  Sul significato di self torniamo tra poco.  

```python
class Animale:
    
    regno = "animale" # attributo di classe 
    
    def __init__(self, nome, eta): # costruttore
        self.nome=nome # attributi di istanza
        self.eta=eta
    
    def __str__(self): # metodo particolare, vedi commento a fine cella.
        return f"Animale di nome {self.nome} di {self.eta} anni."
    
    def presentati(self): # metodo
        print(f"Mi chiamo {self.nome} e ho {self.eta} anni.")



    """FUN FACT
    I metodi di sistema in python sono preceduti e seguiti da due underscores (dunder) e prendono il nome di dunder methods.
    In particolare __str__ specifica il comportamento da seguire quando noi forziamo la conversione a stringa di un oggetto.
    Questo può avvenire quando si esegue print(oggetto) oppure str(oggetto). 
    
    Ci sono dunder methods definire i comportamenti di praticamente tutti gli operatori base (tipo len), operatori matematici (__sum__, __sub__ ...), 
    per le operazioni di flusso (__enter__, __exit__) ecc
    
    """
```

Per creare un oggetto, noi richiamiamo il nome della classe. Richiamare il nome della classe equivale a eseguire la funzione __init__, quindi se la funzione richiede delle variabili, dobbiamo passargliele.  Con questa operazione diciamo che stiamo __istanziando__ la classe.
  
Il "self" non va passato come variabile quando istanziamo la classe, perchè rappresenta l'istanza che stiamo creando.  Questo è necessario per "trasformare" una variabile di una funzione (nome oppure eta) in un attributo del nascituro oggetto della classe.  Questo perchè il modo in cui si assegnano gli attributi ad un oggetto di una classe è proprio il segno =

Qua sopra abbiamo visto come il nostro amico Rex abbia ereditato il costruttore (quindi il metodo in cui viene creato) e il metodo .presentati() dalla superclasse Animle.  

Possiamo anche estendere il costruttore della superclasse con delle aggiunte specifiche della sottoclasse. Vediamo come fare

```python
class Cane(Animale):
    
    specie = "cane"
    
    def __init__(self, nome, eta, razza):
        super().__init__(nome, eta) # chiamo il costruttore della superclasse passandogli le variabili necessarie
        self.razza=razza
    
    def abbaia(self):
        print("Bau!")
    
    def presentati(self):
        super().presentati()  # in questo modo richiamo il metodo della superclasse. Se non lo richiamo, il metodo viene sovrascritto.
        print(f"Sono un {self.razza}") # qua definisco il comportamento aggiuntivo per la sottoclasse
    
```

## Proteggere attributi e metodi

Vogliamo limitare il numero di razze canine possibili. 

In python, qualsiasi attributo è sempre modificabile liberamente. Per convenzione, quando un attributo è limitato e non si deve accedersi liberamente, si mette un underscore _ davanti al nome dell'attributo. La stessa convenzione vale anche per i metodi

```python
class Cane(Animale):
    
    specie = "cane"
    # ricordiamo che gli attributi di classe diventano automaticamente attributi delle istanze
    razze = ["bulldog", "border collie", "labrador", "pitbull"] 
    
    def __init__(self, nome, eta, razza):
        super().__init__(nome, eta)
        if razza not in self.razze:
            raise ValueError("Questa non è una razza canina")
        else: 
            self._razza=razza  # segnalo che l'attributo è "protetto" 
    
    def abbaia(self):
        print("Bau!")
    
    def presentati(self):
        super().presentati()
        print(f"Sono un {self._razza}") 
    
```

```python
# provimo a istanzia
cane_3 = Cane("Akela", 10, "lupo")
```

```python
cane_4 = Cane("Rufus", 2, "labrador")
cane_4.presentati()
```

C'è però un problema: i nostri cani non hanno più un attributo di razza: per proteggerlo ci abbiamo messo un underscore davanti.  

```python
cane_4.razza
```

Per questo motivo esistono i decoratori @property (equivalente a un getter) e @nomeattributo.setter.  


```python
class Cane(Animale):
    
    specie = "cane"
    razze = ["bulldog", "border collie", "labrador", "pitbull"] 
    
    def __init__(self, nome, eta, razza):
        super().__init__(nome, eta)
        if razza not in self.razze:
            raise ValueError("Questa non è una razza canina")
        else: 
            self._razza=razza  

    def abbaia(self):
        print("Bau!")
    
    def presentati(self):
        super().presentati()
        print(f"Sono un {self._razza}") 
        
    @property
    def razza(self):
        return self._razza
    
    @razza.setter
    def razza(self, nuova_razza):
        if nuova_razza not in self.razze:
            raise ValueError("Questa non è una razza canina")
        else:
            self._razza=nuova_razza
    
```

Abbiamo visto che quella di mettere un underscore è una convenzione che rende un metodo protetto, ma niente impedirebbe di sovrascriverlo manualmente.  

Se vogliamo aggiungere un layer di sicurezza aggiuntivo, possiamo renderlo privato con una tecnica chiamata name mangling. Per utilizzarla, definiamo il nuovo attributo "fedeltà". Notiamo che gli underscore ora sono due. 

```python
class Cane(Animale):
    
    specie = "cane"
    razze = ["bulldog", "border collie", "labrador", "pitbull"] 
    
    def __init__(self, nome, eta, razza, fedelta):
        super().__init__(nome, eta)
        if razza not in self.razze:
            raise ValueError("Questa non è una razza canina")
        else: 
            self._razza=razza  
            
        if fedelta not in ["alta", "media", "bassa"]:
            raise ValueError
        else:
            self.__fedelta = fedelta

    def abbaia(self):
        print("Bau!")
    
    def presentati(self):
        super().presentati()
        print(f"Sono un {self._razza}") 
        
    @property
    def razza(self):
        return self._razza
    
    @razza.setter
    def razza(self, nuova_razza):
        if nuova_razza not in self.razze:
            raise ValueError("Questa non è una razza canina")
        else:
            self._razza=nuova_razza
            
    @property
    def fedelta(self):
        return self.__fedelta

    @fedelta.setter
    def fedelta(self, nuova_fedelta):
        if nuova_fedelta not in ["alta", "media", "bassa"]:
            raise ValueError
        self.__fedelta=nuova_fedelta
            
    
```

```python
# per non fare confusione, iniziamo a usare i kwargs
cane_6 = Cane(nome="Balto",
              eta=8, 
              razza="labrador", 
              fedelta="media")

# proviamo ad accedere manualmente (ignorando le convenzioni) all'attributo protetto "razza"
print(cane_6._razza)
```

```python
# proviamo nello stesso modo ad accedere all'attributo privato "fedeltà", ricordandoci che gli underscore sono 2
print(cane_6.__fedelta)
```

L'attributo non viene trovato grazie al name mangling. Quando la classe viene istanziata infatti, l'interprete lo rinomina in `istanza._Classe__attributo`

```python
print(cane_6._Cane__fedelta)
```

Questo offre alcuni spunti interessanti: infatti grazie al name mangling, gli attributi privati non sono sovrascrivibili dalle sottoclassi.  Facciamo un esempio:

```python
class CaneSuperFedele(Cane):
    def __init__(self, nome, eta, razza, fedelta):
        super().__init__(nome, eta, razza, fedelta)
        
        self.__fedelta="altissima" # assegnamo una fedeltà altissima al cane super fedele
        
    def mostra_poteri(self):
        print(f"Fedeltà ereditata : {self._Cane__fedelta}")
        print(f"Fedeltà di questa istanza : {self.__fedelta}")  
        
    def presentati(self):
        super().presentati()
        self.mostra_poteri()
```

```python
cane_7 = CaneSuperFedele(nome="superdog",
                         eta=10, 
                         razza="bulldog", 
                         fedelta="media") # da notare che abbiamo dato una fedelta media al costruttore

cane_7.presentati() 
```

Questo succede perchè l'attributo privato viene salvato con il nome della classe nella sintassi vista prima.

## Metodi che non dipendono dall'istanza: classmethod e staticmethod


Abbiamo visto finora metodi che fanno azioni su una certa istanza, utilizzando "self". Esistono però altri costrutti, che ora vediamo brevemente.

L'utilizzo del decoratore @classmethod inietta la variabile locale "cls" nella funzione del metodo, che ci constente di effettuare azioni sulla classe e non sulla istanza, come ad esempio azionare il costruttore di una classe figlia.
  
L'utilizzo di @staticmethod invece ci consente di creare dei metodi che hanno un funzionamento non collegato alle dinamiche di oop, quindi che non agisce nè su istanze nè sulla classe in generale.  

```python
class Animale:
    
    regno = "animale"  
    
    def __init__(self, nome, eta): 
        self.nome=nome 
        self.eta=eta
    
    def __str__(self): 
        return f"Animale di nome {self.nome} di {self.eta} anni."
    
    def presentati(self): 
        print(f"Mi chiamo {self.nome} e ho {self.eta} anni.")
        
    @classmethod
    def fai_cucciolo(cls, *args, **kwargs):
        return cls(*args, **kwargs, eta=0) # istanzia un nuovo elemento della sottoclasse con età 0
    
    @staticmethod
    def descrizione_generica():
        print("Gli animali sono esseri viventi che popolano la terra.")
    
```

```python
# ricreo la classe Cane utilizzata nelle celle precedente (siccome ho modificato la superclasse). Non ha nessuna modifica rispetto a quanto visto prima
class Cane(Animale):
    
    specie = "cane"
    razze = ["bulldog", "border collie", "labrador", "pitbull"] 
    
    def __init__(self, nome, eta, razza, fedelta):
        super().__init__(nome, eta)
        if razza not in self.razze:
            raise ValueError("Questa non è una razza canina")
        else: 
            self._razza=razza  
            
        if fedelta not in ["media", "bassa", "alta"]:
            raise ValueError
        else:
            self.__fedelta = fedelta

    def abbaia(self):
        print("Bau!")
    
    def presentati(self):
        super().presentati()
        print(f"Sono un {self._razza}") 
        
    @property
    def razza(self):
        return self._razza
    
    @razza.setter
    def razza(self, nuova_razza):
        if nuova_razza not in self.razze:
            raise ValueError("Questa non è una razza canina")
        else:
            self._razza=nuova_razza
            
    @property
    def fedelta(self):
        return self.__fedelta

    @fedelta.setter
    def fedelta(self, nuova_fedelta):
        if nuova_fedelta not in ["media", "bassa", "alta"]:
            raise ValueError
        self.__fedelta=nuova_fedelta
            
```

```python
cane_8 = Cane(nome="Trudi",
              eta=9,
              razza="bulldog",
              fedelta="alta"
)

cucciolo = cane_8.fai_cucciolo(
    nome="Pallino",
    razza="bulldog",
    fedelta="media"
)

cucciolo.presentati()
```

```python
# in conclusione, vediamo anche che pure il "figlio del figlio" ha ereditato il metodo statico comune a tutti gli animali.
cucciolo.descrizione_generica()
```

# Parte 2: esempio pratico 

Abbiamo visto i concetti base, ora vediamo una implementazione forse più "complicata", ma che sicuramente dà un'idea più pratica e realistica degli utilizzi della OOP.  

Utilizzeremo i concetti della prima parte teorica per produrre una rudimentale blockchain di una valuta chimata LMCoin . Una blokchain è una struttura dati decentralizzata e immutabile composta da una catena di blocchi collegati tra loro tramite hash crittografici. Ogni blocco contiene dati (come transazioni), un timestamp, l'hash del blocco precedente e il proprio hash, generato combinando tutte queste informazioni.

Iniziamo quindi progettando la struttura di un blocco.


```python
import time
import hashlib

class Block: # inziamo la definizione del nostro blocco
    
    # partiamo dal cotruttore!
    def __init__(self, index, dati, hash_precedente=''):  # per creare un blocco, avremo bisogno di questi parametri
        self.index = index  # indice, ossia posizione nella nostra bc di questo blocco
        self.timestamp = time.strftime('%Y-%m-%d %H:%M:%S') # stringa con la data di creazione, generata automaticamente alla creazione
        self.dati = dati # contenuto che voglio rendere verificato, ad esempio "TX001: Lorenzo invia a Riccardo 0.4 LMCoin " 
        self.hash_precedente = hash_precedente  # hash precedente nella catena, vediamo meglio dopo cosa significa.
        self.hash_corrente = self.calcola_hash() # servirà da "prova del 9" rispetto all'integrità della struttura
        
    def calcola_hash(self): # funzione che serve a creare l'hash corrente della "sintesi" del contenuto del blocco
        contenuto = f"{self.index}{self.timestamp}{self.dati}{self.hash_precedente}" # questa è la sintesi del contenuto del blocco
        return hashlib.sha256(contenuto.encode()).hexdigest() # <- cripto il mio contenuto di sintesi.
        # PS: questo notebook non parla di crittografia, quindi non approfondiremo il significato della riga qua sopra per non essere dispersivi.
        # Basti sapere che un hash è come l'ombra proiettata di un oggetto: posso sapere con certezza se un'ombra appartiene o meno a un oggetto, ma
        # non posso ricreare un oggetto a partire dalla sua ombra. 
    
    def __str__(self): # definizione della rappresentazione dell'oggetto Block sotto forma di stringa
        return f"Blocco {self.index} [{self.timestamp}] - Dati: {self.dati} - Hash: {self.hash_corrente}"
    
```

Creiamo ora la classe BlockChain. Iniziamo da una versione molto basilare, e poi andremo ad agginugere funzionalità gradualmente.  
In questo esempio, i blocchi rimarranno in cache sotto l'attributo chain.   
Definiamo un comportamento inziale (costruttore con creazione del primo blocco) e alcune funzionalità di base.



```python
class BlockChain:
    def __init__(self): # il costruttore istanzia una blockchain e crea il primo blocco
        self.__chain = [self._crea_primo_blocco()]  # diciamo che la nostra chain sia per ora una lista di blocchi
        
    def _crea_primo_blocco(self): # questo metodo è protetto, perchè non è fatto per essere azionato dall'esterno.
        return Block(0, "Primo Blocco", "0")

    def aggiungi_blocco(self, dati): # metodo per aggiungere un blocco alla nostra chain prendendo solo i dati del blocco
        ultimo_blocco = self.__chain[-1] # recupero l'ultimo blocco
        nuovo_blocco = Block(len(self.__chain), dati, ultimo_blocco.hash_corrente) # creo il blocco 
        self.__chain.append(nuovo_blocco) # aggiuingo tut
        
    def mostra_blocchi(self): # ci servirà anche un metodo per vedere l'intera chain
        for blocco in self.__chain: 
            print(blocco)  # ricordiamoci che abbiamo definito col metodo __str__ il compoortamento di un oggetto Blocco
        
```

```python
# esempio di questa prima versione base
blockchain = BlockChain()
blockchain.aggiungi_blocco("TX001: 10 LMCoin da Lorenzo a Riccardo")
blockchain.aggiungi_blocco("TX002: 5 LMCoin da Giovanna a Ambra")
blockchain.aggiungi_blocco("TX003: 2 LMCoin da Ablerto a Davide")
blockchain.mostra_blocchi()
```

Bene, siamo in grado di inserire blocchi nella nostra struttura dati e di visualizzarne il contenuto. Ma siamo ancora lontani dal creare una valida alternativa al denaro contante!  

Iniziamo a pensare alla sicurezza, e implementiamo un metodo che consenta di vedere se un blocco è stato alterato.

```python
# NOTA IMPORTANTE! FINO AL METODO mostra_blocchi() ho copiato la cella di codice precedente
# E' indispensabile che ogni volta che aggiungo metodi alla classe la ri-definisca nella sua interezza

class BlockChain:
    def __init__(self): 
        self.__chain = [self._crea_primo_blocco()]   
        
    def _crea_primo_blocco(self):  
        return Block(0, "Primo Blocco", "0")

    def aggiungi_blocco(self, dati):  
        ultimo_blocco = self.__chain[-1]  
        nuovo_blocco = Block(len(self.__chain), dati, ultimo_blocco.hash_corrente)  
        self.__chain.append(nuovo_blocco)  
        
    def mostra_blocchi(self):  
        for blocco in self.__chain: 
            print(blocco)   
    
    def controlla_integrita(self): # iniziamo la dichiarazione del metodo che ci consente di controllare l integrità della nostra blockchain
               
        for i in range(1, len(self.__chain)): # itero su ciascun blocco dopo quello iniziale (che ha indice 0)
            
            blocco_corrente = self.__chain[i] # assegno il blocco corrispondente all'iterazione alla variabile blocco_corrente ...
            blocco_precedente = self.__chain[i - 1] # ... e il suo precedente alla variabile blocco_precedente

            # Controllo innanzitutto se ci sono differenze tra l'hash salvato sul blocco come attributo e quello che ottengo ricalcolandolo.
            # Comprensibilmente, mi aspetto che siano uguali. Se non lo sono, significa che ci sono state alterazioni
            if blocco_corrente.hash_corrente != blocco_corrente.calcola_hash(): 
                print(f"Errore: Il blocco {i} è stato compromesso!")            
                return 
            
            # Ora che so che il blocco_corrente è valido, faccio lo stesso controllo di consistenza tra l' "hash_precedente" salvato come attributo
            # sul blocco corrente (che abbiamo stabilito essere valido) e quello salvato sul blocco precedente come "hash_corrente".
            # Se non corrispondono, significa che è stato aggiunto un blocco "taroccato" in mezzo a due blocchi validi
            if blocco_corrente.hash_precedente != blocco_precedente.hash_corrente: 
                print(f"Errore: Il blocco {i} non è consistente con il precednete!")    
                return                                                        
        print("Blockchain valida!")
        return 
        
```

```python
blockchain = BlockChain()
blockchain.aggiungi_blocco("TX001: 10 LMCoin da Lorenzo a Riccardo")
blockchain.aggiungi_blocco("TX002: 5 LMCoin da Giovanna a Ambra")
blockchain.aggiungi_blocco("TX003: 2 LMCoin da Ablerto a Davide")
blockchain.mostra_blocchi()
```

```python
blockchain.controlla_integrita()
```
